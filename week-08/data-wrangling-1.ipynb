## What to keep

- **loc.** Keeps rows that pass a test. The original table still has every penguin. A missing number fails a comparison.
- **A list of names.** Keeps those columns. The other columns stay in the original table.
- **sort_values.** Orders the rows. `ascending=False` puts the largest first. Missing values go to the end.
- **value_counts and groupby.** Count rows. `dropna=False` keeps a blank in the count.
- **A new column.** Compute it on a copy. The columns it came from stay as they were.
